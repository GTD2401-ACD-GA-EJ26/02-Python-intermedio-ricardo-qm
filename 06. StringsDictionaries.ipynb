```python
x = 'Pluto is a planet'
y = "Pluto is a planet"
x == y
```

```python
hello = "hello\nworld"
print(hello)
```

```python
triplequoted_hello = """hello
world"""
print(triplequoted_hello)
triplequoted_hello == hello
```

```python
print("hello")
print("world")
print("hello", end='')
print("pluto", end='')
```

```python
# Indexing
planet = 'Pluto'
planet[0]
```

```python
# Yes, we can even loop over them
[char+'! ' for char in planet]
```

```python
planet[0] = 'B'
# planet.append doesn't work either
```

```python
# ALL CAPS
claim = "Pluto is a planet!"
claim.upper()
```

```python
words = claim.split()
words
```

```python
datestr = '1956-01-31'
year, month, day = datestr.split('-')
```

```python
'/'.join([month, day, year])
```

```python
# Yes, we can put unicode characters right in our string literals :)
' 👏 '.join([word.upper() for word in words])
```
